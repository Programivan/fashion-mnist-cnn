# Класифікація одягу за допомогою згорткової нейронної мережі (CNN) на Fashion MNIST

**Мета завдання:** Створити згорткову нейронну мережу (CNN) для класифікації зображень одягу з набору даних Fashion MNIST. Дослідити вплив різних шарів (Conv2D, MaxPooling, Dropout, Dense) та кількості нейронів/фільтрів на точність класифікації, а також побудувати графіки точності та похибки за епохами.

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

print(f"TensorFlow версія: {tf.__version__}")
print("Доступні пристрої:", tf.config.list_physical_devices())
```

## 1. Завантаження та попередня обробка даних Fashion MNIST

```python
class_names = [
    "T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
    "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"
]

fashion_mnist = tf.keras.datasets.fashion_mnist
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# Нормалізація до [0, 1] та додавання виміру каналу
X_train = (X_train.astype(np.float32) / 255.0).reshape(-1, 28, 28, 1)
X_test = (X_test.astype(np.float32) / 255.0).reshape(-1, 28, 28, 1)

print("Форма тренувальних даних:", X_train.shape, y_train.shape)
print("Форма тестових даних:", X_test.shape, y_test.shape)
```

## 2. Створення архітектури CNN з різними шарами та нейронами

```python
model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
model.summary()
```

## 3. Навчання нейромережі

```python
history = model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=64,
    validation_data=(X_test, y_test),
    verbose=1
)
```

## 4. Побудова графіків Accuracy та Loss (як у завданні)

```python
plt.figure(figsize=(11, 4.5))

# Графік точності
plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Train accuracy", color="#1f77b4")
plt.plot(history.history["val_accuracy"], label="Val accuracy", color="#ff7f0e")
plt.title("Accuracy over epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(loc="upper left")

# Графік втрат
plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Train loss", color="#1f77b4")
plt.plot(history.history["val_loss"], label="Val loss", color="#ff7f0e")
plt.title("Loss over epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend(loc="upper right")

plt.tight_layout()
plt.show()
```

## 5. Оцінка моделі на тестових даних

```python
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"Точність на тестовій вибірці: {test_acc * 100:.2f}%")
print(f"Похибка на тестовій вибірці: {test_loss:.4f}")
```
